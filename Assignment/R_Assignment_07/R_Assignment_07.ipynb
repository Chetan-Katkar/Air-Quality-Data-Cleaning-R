# Lecture 28 — Repeats

## 19. Help for `rep`

- `rep(x)` — replicates the values in x
- `rep(x, times = n)` — repeat x as a whole n times
- `rep(x, each = n)` — repeat each cell n times
- `rep(x, length.out = n)`, `rep(x, length = n)`, `rep_len(x, length.out)`

```r
help("rep")
args(rep)
```

## 20. Repeat an object n times

```r
rep(3.5, times = 10)
rep(1:4, 2)
```

## 21. `times` versus `each`

```r
x = 1:4
x
rep(x, times = 3)
rep(x, each = 3)
```

## 22. Effect of the ordering of `each` and `times`

Both give the same result — R matches the arguments by name, not by position.

```r
rep(1:4, each = 2)
rep(1:4, each = 2, times = 3)
rep(1:4, times = 3, each = 2)
```

## 23. Every object repeated a different number of times

The `times` argument can itself be a vector of the same length as x.

```r
rep(1:4, 2:5)
```

```r
ans = seq(from = 2, to = 8, by = 2)
ans
rep(1:4, ans)
```

## 24. Repeating a matrix

A matrix is stored column-wise, so `rep` flattens it column by column.

```r
x = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = T)
x
rep(x, 3)
```

## 25. Repetition of characters

```r
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)
```

## 26. Repetition for a pre-specified length

```r
rep(2, length.out = 5)
rep(2, length = 5)
rep(c(2, 3), length = 5)
rep(c(2, 3, 4), length = 5)
```

```r
rep("apple", length = 5)
rep(c("a", "b", "c"), length = 2)
rep(c("a", "b", "c"), length = 5)
```

# Lecture 29 — Sorting, Ordering and Mode

## 27. Sorting

`sort(x, decreasing = FALSE, na.last = NA)`

- `na.last = TRUE` — missing values placed last
- `na.last = FALSE` — missing values placed first
- `na.last = NA` — missing values removed

```r
y = c(8, 5, 7, 6)
y
sort(y)
sort(y, decreasing = TRUE)
```

## 28. Sorting with missing values

```r
ymiss = c(8, NA, 5, 7, NA, 6)
sort(ymiss)                  # na.last = NA, NAs dropped
sort(ymiss, na.last = TRUE)  # NAs pushed to the end
sort(ymiss, na.last = FALSE) # NAs pulled to the front
```

## 29. Ordering

`order` returns the **positions** that would sort the vector.

| Value | 9 | 8 | 5 | 7 | 6 |
|---|---|---|---|---|---|
| Position | 1 | 2 | 3 | 4 | 5 |

| Ordered value | 5 | 6 | 7 | 8 | 9 |
|---|---|---|---|---|---|
| Position | 3 | 5 | 4 | 2 | 1 |

```r
y = c(9, 8, 5, 7, 6)
y
order(y)
order(y, decreasing = TRUE)
```

```r
# order() used to actually rearrange the vector
y[order(y)]
```

## 30. Mode

Every object has a mode, which indicates how the object is stored in memory.

```r
mode(2.432)
mode(c(3, 4, 5, 6, 7, 8))
mode("India")
mode(c("India", "CANADA"))
```

```r
mode(factor(c("MP", "UP")))
mode(list("India", "CANADA"))
mode(data.frame(x = 1:2, y = c("India", "CANADA")))
mode(print)
```

# Lecture 30 — Lists

## 31. Vectors and matrices hold only one type of data

A list, in contrast, can contain elements of different modes, can hold structured
objects such as data frames, can be indexed by position (`x[[5]]`), can be
sub-setted (`x[c(2,5)]`), and its elements can be named
(`x[["Students"]]` or `x$Students`).

```r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = T)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = T)
x1
x2
x1 + x2
```

## 32. A matrix cannot mix modes

Assigning a character into a numeric matrix coerces the whole matrix to character,
and arithmetic then fails.

`try()` is used here only so that the error message is printed without stopping
the notebook.

```r
x1[2, 1] = "hello"
x1
try(x1 + x2)
```

## 33. Lists can contain any kind of object, including matrices

```r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = T)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = T)
matlist = list(x1, x2)
matlist
```

## 34. `[ ]` returns a sublist, `[[ ]]` returns the element itself

```r
matlist[1]
matlist[2]
```

```r
# Compare with the double bracket
matlist[[1]]
mode(matlist[1])    # still a list
mode(matlist[[1]])  # the matrix itself
```

## 35. A list containing different object types

```r
z1 = list( c("water", "juice", "lemonade"),
           rep(1:4, each = 2),
           matrix(data = 5:8, nrow = 2, ncol = 2, byrow = T) )
z1
```

## 36. Accessing elements of a list

`z1[1][2]` returns NULL, while `z1[[1]][2]` returns the desired element.

```r
z1[[1]]
z1[1][2]     # NULL - wrong bracket
z1[[1]][2]   # "juice" - correct
```

## 37. Named list elements

`x[["Students"]]` and `x$Students` refer to the same element.

```r
x = list(Students = c("Amit", "Rekha", "Swaroop"),
         Marks = c(88, 91, 79),
         Course = "R Programming")
x
x[["Students"]]
x$Students
```

```r
# Sublist of selected elements
x[c(1, 3)]
names(x)
length(x)
```

## Quick recap

| Task | Command |
|---|---|
| Unit-increment sequence | `1:10`, `-5:-15` |
| General sequence | `seq(from, to, by, length.out)` |
| Index vector | `seq(along = x)` |
| Current date / time | `Sys.Date()`, `Sys.time()` |
| Date sequence | `seq(as.Date(...), by = "months", length = 6)` |
| Alphabets | `letters`, `LETTERS` |
| Repeat whole object | `rep(x, times = n)` |
| Repeat each element | `rep(x, each = n)` |
| Repeat to a fixed length | `rep(x, length.out = n)` |
| Sort values | `sort(x, decreasing, na.last)` |
| Sort positions | `order(x)` |
| Storage type | `mode(x)` |
| Mixed-type container | `list(...)`, `x[[i]]`, `x$name` |
